```python
# calculate the probability of a driver scoring above, below, or equal his Fantasy points average

import matplotlib.pyplot as plt
import numpy as np

driver = input('Driver: ').upper()
data = list(input(f'Points progression ({driver}): ').split())
data = np.array(list(map(int, data)))
avg = np.mean(data)

over = 0
equal = 0
under = 0
for num in data:
  if num > avg:
    over += 1
  elif num == avg:
    equal += 1
  elif num < avg:
    under += 1
over_pct = (over/len(data)) * 100
equal_pct = (equal/len(data)) * 100
under_pct = (under/len(data)) * 100

categories = ['Over AVG', 'Equal AVG', 'Under AVG']
values = [round(over_pct, 1), round(equal_pct, 1), round(under_pct, 1)]
fig, ax = plt.subplots()
bars = ax.bar(categories, values, color='blue')
ax.bar_label(bars, label_type='center', color='white', fontsize=12, weight='bold')
plt.ylim(0, 100)
plt.title(f'Average Fantasy Points ({driver}): {round(avg, 1)}')
plt.xlabel('Tendency to Score')
plt.ylabel('Percentage (%)')
plt.show()
```

```python
# calculate the probability of a driver scoring in a range of points

import pandas as pd
import matplotlib.pyplot as plt

def zero_or_lesser(data):
  count = 0
  for entry in data:
    if entry <= 0:
      count += 1
  result = (count / len(data)) * 100
  return round(result, 1)

def mid_range(min_num, max_num, data):
  count = 0
  for entry in data:
    if min_num <= entry <= max_num:
      count += 1
  result = (count / len(data)) * 100
  return round(result, 1)

def hundred_plus(data):
  count = 0
  for entry in data:
    if entry > 100:
      count += 1
  result = (count / len(data)) * 100
  return round(result, 1)

driver = input('Driver: ').upper()
points = list(input(f'Points progression ({driver}): ').split())
points = list(map(int, points))

values = []
categories = []
values.append(zero_or_lesser(points))
categories.append('<= 0')

start = 1
stop = 10
while stop <= 100:
  values.append(mid_range(start, stop, points))
  categories.append(f'{start}-{stop}')
  start += 10
  stop += 10

values.append(hundred_plus(points))
categories.append('> 100')

fig, ax = plt.subplots()
bars = ax.barh(categories, values, color='red')
ax.bar_label(bars, padding=5, fontsize=10, fontweight='bold')
ax.set_xlim(0, max(values) * 1.1)
plt.xlabel('Probability of Scoring (%)')
plt.ylabel('Scoring Range')
plt.show()
```
