```python
# calculate total team value if future updates are factored in

values = []
updates = []
for num in range(1, 6):
  driver = input(f"Driver {num}: ").upper()
  value = float(input(f'{driver} value: '))
  values.append(value)
  update = float(input('Value update: '))
  updates.append(update)
for num in range(1, 3):
  constructor = input(f'Constructor {num}: ').upper()
  value = float(input(f'{constructor} value: '))
  values.append(value)
  update = float(input('Value update: '))
  updates.append(update)

cost_cap = float(input('Cost cap (M): '))
team_value = sum(values)
total_updates = sum(updates)
print('-' * 8)
print(f'Value of selected team: {round(team_value, 1)}')
print(f'Value of selected team + cost cap: {round(team_value + cost_cap, 1)}')
print(f'Total updates: {round(total_updates, 1)}')
print(f'Updated value of selected team: {round(team_value + total_updates, 1)}')
print(f'Updated value of selected team + cost cap: {round(team_value + total_updates + cost_cap, 1)}')
```

```python
# calculate the number of races a driver needs to reach Tier A or go down to Tier B

driver = input('Driver: ').upper()
value = float(input(f'{driver} value (M): '))
if value < 18.5:
  upgrade = float(input('Select constant upgrade (0.6 or 0.2): '))
  races = 0
  while value < 18.5:
    value += upgrade
    races += 1
  print(f'Races needed to get to Tier A: {races}')
elif value >= 18.5:
  downgrade = float(input('Select constant downgrade (0.3 or 0.1): '))
  races = 0
  while value >= 18.5:
    value -= downgrade
    races += 1
  print(f'Races needed to go down to Tier B: {races}')
```

```python
# calculate the probability of a driver scoring above, below, or equal his Fantasy points average

import matplotlib.pyplot as plt
import numpy as np

driver = input('Driver: ').upper()
data = list(input(f'Points progression ({driver}): ').split())
data = np.array(list(map(int, data)))
avg = np.mean(data)

over = 0
equal = 0
under = 0
for num in data:
  if num > avg:
    over += 1
  elif num == avg:
    equal += 1
  elif num < avg:
    under += 1
over_pct = (over/len(data)) * 100
equal_pct = (equal/len(data)) * 100
under_pct = (under/len(data)) * 100

categories = ['Over AVG', 'Equal AVG', 'Under AVG']
values = [round(over_pct, 1), round(equal_pct, 1), round(under_pct, 1)]
fig, ax = plt.subplots()
bars = ax.bar(categories, values, color='blue')
ax.bar_label(bars, label_type='center', color='white', fontsize=12, weight='bold')
plt.ylim(0, 100)
plt.title(f'Average Fantasy Points ({driver}): {round(avg, 1)}')
plt.xlabel('Tendency to Score')
plt.ylabel('Percentage (%)')
plt.show()
```
