```python
import numpy as np
import pandas as pd

drivers = []
avg_pts = []
dev_scores = []
min_pts = []
med_pts = []
max_pts = []
total_drivers = int(input('Number of drivers: '))
for n in range(1, total_drivers + 1):
  driver = input(f'Driver {n}: ').upper()
  drivers.append(driver)
  points = input('Points progression: ')
  array = np.fromstring(points, dtype=int, sep=' ')
  avg_pts.append(np.mean(array))
  dev_scores.append(np.std(array))
  min_pts.append(np.min(array))
  med_pts.append(np.median(array))
  max_pts.append(np.max(array))

data = {'driver': [i for i in drivers],
        'avg_pts': [i for i in avg_pts]}
df = pd.DataFrame(data)
df_sorted = df.sort_values('avg_pts', ascending=False)
print('-' * 5)
print(df_sorted.to_string())
```

```python
import pandas as pd

data = {'driver': [i for i in drivers],
        'dev_score': [i for i in dev_scores]}
df = pd.DataFrame(data)
df_sorted = df.sort_values('dev_score')
print(df_sorted.to_string())
```

```python
import pandas as pd

data = {'driver': [i for i in drivers],
        'min': [i for i in min_pts],
        'med': [i for i in med_pts],
        'max': [i for i in max_pts]}
df = pd.DataFrame(data)
df_sorted = df.sort_values('med', ascending=False)
print(df_sorted.to_string())
```

```python
import matplotlib.pyplot as plt

driver1 = input('Driver 1: ').upper()
driver1_data = list(input(f'Points progression ({driver1}): ').split())
driver1_data = list(map(int, driver1_data))
color1 = input(f'Line color ({driver1}): ')

driver2 = input('Driver 2: ').upper()
driver2_data = list(input(f'Points progression ({driver2}): ').split())
driver2_data = list(map(int, driver2_data))
color2 = input(f'Line color ({driver2}): ')

x = [n for n in range(1, len(driver1_data) + 1)]
y1 = [n for n in driver1_data]
y2 = [n for n in driver2_data]

plt.plot(x, y1, label=driver1, color=color1)
plt.plot(x, y2, label=driver2, color=color2)

plt.xlabel('Race')
plt.ylabel('Season Form')
plt.legend()
plt.show()
```
