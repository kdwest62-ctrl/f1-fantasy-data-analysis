```python
import matplotlib.pyplot as plt

drivers = []
points = []
total = int(input("Number of drivers: "))
for num in range(1, total + 1):
  driver = input(f"Driver {num}: ").upper()
  drivers.append(driver)
  total_points = sum(driver_pts[driver])
  points.append(total_points)

plt.bar(drivers, points)
plt.xlabel("Driver")
plt.ylabel("Total Fantasy Points")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

driver1 = input('Driver 1: ').upper()
driver1_pts = np.array(driver_pts[driver1])
result1 = np.std(driver1_pts)

driver2 = input('Driver 2: ').upper()
driver2_pts = np.array(driver_pts[driver2])
result2 = np.std(driver2_pts)

x_axis = [driver1, driver2]
y_axis = [result1, result2]

plt.barh(x_axis, y_axis)
plt.xlabel('Standard Deviation')
plt.ylabel('Driver')
plt.show()
```

```python
import matplotlib.pyplot as plt

driver1 = input('Driver 1: ').upper()
driver1_pts = driver_pts[driver1]
color1 = input('Line color (Driver 1): ')

driver2 = input('Driver 2: ').upper()
driver2_pts = driver_pts[driver2]
color2 = input('Line color (Driver 2): ')

x = [n for n in range(1, len(driver1_pts) + 1)]
y1 = [n for n in driver1_pts]
y2 = [n for n in driver2_pts]

plt.plot(x, y1, label=driver1, color=color1)
plt.plot(x, y2, label=driver2, color=color2)

plt.xlabel('Race')
plt.ylabel('Fantasy Points Progression')
plt.legend()
plt.show()
```
