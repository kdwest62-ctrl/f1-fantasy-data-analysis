```python
# compare 2 or more drivers' total fantasy points

import matplotlib.pyplot as plt

drivers = []
points_per_driver = []
total_drivers = int(input("Number of drivers: "))
for num in range(1, total_drivers + 1):
  driver = input(f"Driver {num}: ").upper()
  drivers.append(driver)

for driver in drivers:
  pts_progress = list(input(f'Points progression ({driver}): ').split())
  total_points = sum(list(map(int, pts_progress)))
  points_per_driver.append(total_points)

fig, ax = plt.subplots()
bars = ax.bar(drivers, points_per_driver, color='red')
ax.bar_label(bars, padding=3, label_type='center', color='white', fontsize=12, fontweight='bold')
plt.xlabel('Driver')
plt.ylabel('Total Fantasy Points')
plt.show()
```
