```python
# compare 2 or more drivers' total fantasy points

import matplotlib.pyplot as plt

drivers = []
points_per_driver = []
total_drivers = int(input("Number of drivers: "))
for num in range(1, total_drivers + 1):
  driver = input(f"Driver {num}: ").upper()
  drivers.append(driver)

for driver in drivers:
  data = list(input(f'Points progression ({driver}): ').split())
  data = list(map(int, data))
  points_per_driver.append(sum(data))

fig, ax = plt.subplots()
bars = ax.bar(drivers, points_per_driver, color='red')
ax.bar_label(bars, padding=3, label_type='center', color='white', fontsize=12, fontweight='bold')
plt.xlabel('Driver')
plt.ylabel('Total Fantasy Points')
plt.show()
```

```python
# compare 2 or more drivers' average fantasy points

import matplotlib.pyplot as plt

drivers = []
avgs = []
total_drivers = int(input("Number of drivers: "))
for num in range(1, total_drivers + 1):
  driver = input(f"Driver {num}: ").upper()
  drivers.append(driver)

for driver in drivers:
  data = list(input(f'Points progression ({driver}): ').split())
  data = list(map(int, data))
  avg = sum(data) / len(data)
  avgs.append(avg)

fig, ax = plt.subplots()
bars = ax.bar(drivers, avgs, color='red')
ax.bar_label(bars, padding=3, label_type='center', color='white', fontsize=12, fontweight='bold')
plt.xlabel('Driver')
plt.ylabel('Avg Fantasy Points per Race')
plt.show()
```

```python
# compares standard deviation scores of 2 or more drivers
# standard deviation measures consistency. The closer the score to zero, the more consistent the scoring is.

import numpy as np
import matplotlib.pyplot as plt

drivers = []
dev_scores = []
total_drivers = int(input("Number of drivers: "))
for num in range(1, total_drivers + 1):
  driver = input(f"Driver {num}: ").upper()
  drivers.append(driver)

for driver in drivers:
  data = input(f'Points progression ({driver}): ')
  array = np.fromstring(data, dtype=int, sep=" ")
  dev_scores.append(np.std(array))

fig, ax = plt.subplots()
bars = ax.bar(drivers, dev_scores, color='red')
ax.bar_label(bars, padding=3, label_type='center', color='white', fontsize=12, fontweight='bold')
plt.xlabel('Driver')
plt.ylabel('Standard Deviation Score')
plt.show()
```

```python
# compare fantasy points progression of 2 drivers

import matplotlib.pyplot as plt

driver1 = input('Driver 1: ').upper()
driver1_data = list(input(f'Points progression ({driver1}): ').split())
driver1_data = list(map(int, driver1_data))
color1 = input(f'Line color ({driver1}): ')

driver2 = input('Driver 2: ').upper()
driver2_data = list(input(f'Points progression ({driver2}): ').split())
driver2_data = list(map(int, driver2_data))
color2 = input(f'Line color ({driver2}): ')

x = [n for n in range(1, len(driver1_data) + 1)]
y1 = [n for n in driver1_data]
y2 = [n for n in driver2_data]

plt.plot(x, y1, label=driver1, color=color1)
plt.plot(x, y2, label=driver2, color=color2)

plt.xlabel('Race')
plt.ylabel('Season Form')
plt.legend()
plt.show()
```
