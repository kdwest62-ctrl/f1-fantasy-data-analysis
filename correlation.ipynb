```python
import matplotlib.pyplot as plt

competitor1 = input('Competitor 1: ').upper()
dataset1 = list(input('Points progression: ').split())
dataset1 = list(map(int, dataset1))

competitor2 = input('Competitor 2: ').upper()
dataset2 = list(input('Points progression: ').split())
dataset2 = list(map(int, dataset2))

plt.scatter(dataset1, dataset2, color='red')
plt.xlabel(competitor1)
plt.ylabel(competitor2)
plt.show()
```

```python
import matplotlib.pyplot as plt

constructor = input('Constructor: ').upper()
constructor_data = list(input(f'Points progression ({constructor}): ').split())
constructor_data = list(map(int, constructor_data))

driver1 = input('Driver 1: ').upper()
driver1_data = list(input(f'Points progression ({driver1}): ').split())
driver1_data = list(map(int, driver1_data))
color1 = input('Marker color: ')

driver2 = input('Driver 2: ').upper()
driver2_data = list(input(f'Points progression ({driver2}): ').split())
driver2_data = list(map(int, driver2_data))
color2 = input('Marker color: ')

fig, ax1 = plt.subplots(figsize=(10, 6))
ax1.scatter(constructor_data, driver1_data, color=color1, label=driver1)
ax1.set_xlabel(f'{constructor} Points')
ax1.set_ylabel(f'{driver1} Points')
ax1.tick_params(axis='y')

ax2 = ax1.twinx()
ax2.scatter(constructor_data, driver2_data, color=color2, label=driver2)
ax2.set_ylabel(f'{driver2} Points')
ax2.tick_params(axis='y')

plt.title(f'{constructor} Points vs. {driver1} and {driver2} Points Progression')
ax1.grid(True)

fig.legend(loc='upper left', bbox_to_anchor=(0.1, 0.9))
plt.show()
```
