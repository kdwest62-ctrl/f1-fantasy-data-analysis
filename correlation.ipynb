```python
# shows correlation between 2 drivers' fantasy points progression

import matplotlib.pyplot as plt

driver1 = input('Driver 1: ').upper()
pts_progress1 = list(input(f'Points progression ({driver1}): ').split())
pts_progress1 = list(map(int, pts_progress1))

driver2 = input('Driver 2: ').upper()
pts_progress2 = list(input(f'Points progression ({driver2}): ').split())
pts_progress2 = list(map(int, pts_progress2))

plt.scatter(pts_progress1, pts_progress2, color='red')
plt.xlabel(driver1)
plt.ylabel(driver2)
plt.grid()
plt.show()
```

```python
# shows correlation between a driver's and constructor's fantasy points progression

import matplotlib.pyplot as plt

driver = input('Driver: ').upper()
driver_data = list(input(f'Points progression ({driver}): ').split())
driver_data = list(map(int, driver_data))

constructor = input('Constructor: ').upper()
constructor_data = list(input(f'Points progression ({constructor}): ').split())
constructor_data = list(map(int, constructor_data))

plt.scatter(driver_data, constructor_data, color='red')
plt.xlabel(driver)
plt.ylabel(constructor)
plt.grid()
plt.show()
```
