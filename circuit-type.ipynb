```
# legend: L for low-speed, M for medium-speed, H for high-speed

circuits = {
    "AUS": "M",  # Australian GP
    "CHN": "M",  # Chinese GP
    "JPN": "M",  # Japanese GP
    "MIA": "M",  # Miami GP
    "CAN": "M",  # Canadian GP
    "MON": "L",  # Monaco GP
    "BAR": "M",  # Barcelona-Catalunya GP
    "AUT": "H",  # Austrian GP
    "GBR": "H",  # British GP
    "BEL": "H",  # Belgian GP
    "HUN": "L",  # Hungarian GP
    "NED": "M",  # Dutch GP
    "ITA": "H",  # Italian GP
    "MAD": "M",  # Madrid Street Circuit
    "AZE": "H",  # Azerbaijan GP
    "BHR": "H",  # Bahrain GP in Malaysia
    "SGP": "L",  # Singapore GP
    "USA": "M",  # United States GP
    "MEX": "H",  # Mexico City GP
    "BRA": "M",  # Sao Paulo GP
    "LVG": "H",  # Las Vegas GP
    "QAT": "M",  # Qatar GP
    "ABU": "M"   # Abu Dhabi GP
}
```

```
# Calculates a driver's Fantasy points distribution per circuit type

import matplotlib.pyplot as plt
import itertools

driver = input('Driver: ').upper()
data = list(input(f'Points progression ({driver}): ').split())
data = list(map(int, data))
num_races = len(data)
races = dict(itertools.islice(circuits.items(), num_races))

low = []
medium = []
high = []
index = 0
for circuit_type in races.values():
  if circuit_type == 'L':
    points = data[index]
    low.append(points)
    index += 1
  elif circuit_type == 'M':
    points = data[index]
    medium.append(points)
    index += 1
  elif circuit_type == 'H':
    points = data[index]
    high.append(points)
    index += 1

categories = [f'Low (races: {len(low)})',
              f'Medium (races: {len(medium)})',
              f'High (races: {len(high)})']
values = [sum(low), sum(medium), sum(high)]
fig, ax = plt.subplots()
bars = ax.bar(categories, values, color='darkblue')
ax.bar_label(bars, label_type='center', color='white', fontsize=12, fontweight='bold')
plt.title(f'Total Fantasy Points ({driver})')
plt.xlabel('Circuit Type (Speed)')
plt.ylabel('Fantasy Points')
plt.show()
```
