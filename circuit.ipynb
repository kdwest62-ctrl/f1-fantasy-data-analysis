```
circuits = {
    "AUS": "M",  # Australian GP
    "CHN": "M",  # Chinese GP
    "JPN": "M",  # Japanese GP
    "MIA": "M",  # Miami GP
    "CAN": "M",  # Canadian GP
    "MON": "L",  # Monaco GP
    "BAR": "M",  # Barcelona-Catalunya GP
    "AUT": "H",  # Austrian GP
    "GBR": "H",  # British GP
    "BEL": "H",  # Belgian GP
    "HUN": "L",  # Hungarian GP
    "NED": "M",  # Dutch GP
    "ITA": "H",  # Italian GP
    "MAD": "M",  # Madrid Street Circuit
    "AZE": "H",  # Azerbaijan GP
    "BHR": "H",  # Bahrain GP in Malaysia
    "SGP": "L",  # Singapore GP
    "USA": "M",  # United States GP
    "MEX": "H",  # Mexico City GP
    "BRA": "M",  # Sao Paulo GP
    "LVG": "H",  # Las Vegas GP
    "QAT": "M",  # Qatar GP
    "ABU": "M"   # Abu Dhabi GP
}
```

```
import itertools
import pandas as pd

competitor = input('Competitor: ').upper()
points = list(input('Points progression: ').split())
points = list(map(int, points))

races = dict(itertools.islice(circuits.items(), len(points)))
data = {'circuit': [i for i in races.keys()], 'type': [i for i in races.values()], 'points': [i for i in points]}
df = pd.DataFrame(data)
print('-' * 5)
print(df.to_string())
print(f'Competitor: {competitor}')
```

```
import matplotlib.pyplot as plt

low = []
medium = []
high = []
index = 0
for circuit_type in races.values():
  if circuit_type == 'L':
    entry = points[index]
    low.append(entry)
    index += 1
  elif circuit_type == 'M':
    entry = points[index]
    medium.append(entry)
    index += 1
  elif circuit_type == 'H':
    entry = points[index]
    high.append(entry)
    index += 1

categories = [f'Low (races: {len(low)})',
              f'Medium (races: {len(medium)})',
              f'High (races: {len(high)})']
values = [sum(low), sum(medium), sum(high)]

fig, ax = plt.subplots()
bars = ax.bar(categories, values, color='darkblue')
ax.bar_label(bars, label_type='center', color='white', fontsize=12, fontweight='bold')
plt.title(f'Competitor: {competitor}')
plt.xlabel('Circuit Type (Speed)')
plt.ylabel('Total Fantasy Points')
plt.show()
```

```
import matplotlib.pyplot as plt
import numpy as np

low_arr = np.array(low)
med_arr = np.array(medium)
high_arr = np.array(high)
categories = [f'Low (races: {len(low)})',
              f'Medium (races: {len(medium)})',
              f'High (races: {len(high)})']

low_mean = np.mean(low_arr) if low_arr.size > 0 else 0.0
med_mean = np.mean(med_arr) if med_arr.size > 0 else 0.0
high_mean = np.mean(high_arr) if high_arr.size > 0 else 0.0
values = [low_mean, med_mean, high_mean]

fig, ax = plt.subplots()
bars = ax.bar(categories, values, color='darkblue')
ax.bar_label(bars, label_type='center', color='white', fontsize=12, fontweight='bold')
plt.title(f'Competitor: {competitor}')
plt.xlabel('Circuit Type (Speed)')
plt.ylabel('Average Fantasy Points')
plt.show()
```

```
import matplotlib.pyplot as plt
import numpy as np

categories = [f'Low (races: {len(low_arr)})',
              f'Medium (races: {len(med_arr)})',
              f'High (races: {len(high_arr)})']

low_std = np.std(low_arr) if low_arr.size > 0 else 0.0
med_std = np.std(med_arr) if med_arr.size > 0 else 0.0
high_std = np.std(high_arr) if high_arr.size > 0 else 0.0
values = [low_std, med_std, high_std]

fig, ax = plt.subplots()
bars = ax.bar(categories, values, color='darkblue')
ax.bar_label(bars, label_type='center', color='white', fontsize=12, fontweight='bold')
plt.title(f'Competitor: {competitor}')
plt.xlabel('Circuit Type (Speed)')
plt.ylabel('Standard Deviation')
plt.show()
```
